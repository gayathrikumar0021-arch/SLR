```python
import pandas as pd
```

```python
dataset=pd.read_csv("Salary_dataset.csv")
```

```python
dataset
```

```python
dataset=dataset.drop(columns=['Unnamed: 0'])
```

```python
dataset
```

```python
dataset.isnull()
```

```python
dataset.isnull().sum()
```

```python
import matplotlib.pyplot as plt
```

```python
x=dataset[["YearsExperience"]]
y=dataset[["Salary"]]
```

```python
plt.scatter(x,y)
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=11)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
splitting=LinearRegression() 
splitting.fit(x_train,y_train)
```

```python
y_pred=splitting.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
r2score=r2_score(y_test,y_pred)
```

```python
r2score
```

```python

import pickle
filename="Myproject.sav"
pickle.dump(splitting,open(filename,'wb'))
```
